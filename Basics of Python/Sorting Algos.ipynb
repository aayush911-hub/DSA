## Bubble Sort

- Compare the adjacent elements and swap the bigger value with the smaller one.

- Time Complexity

    - Best Case: *O(N)*
    - Worst Case: *O(N<sup>2</sup>)*
    - Average Case: *O(N<sup>2</sup>)*


- Space Complexity: O(1)

```python
def bubble_sort(lst):
    size= len(lst)
    n=size
    for i in range(size):
        swap= False
        for j in range(n-1):
            if lst[j]>lst[j+1]:
                lst[j], lst[j+1]= lst[j+1], lst[j]
                swap=True

        if not swap:
            break
        n-=1
    return lst
```

```python
lst= [8,6,4,2,7,9]
sorted_lst= bubble_sort(lst)
print(sorted_lst)
```

## Selection Sort

- Compare the current element with minimum element  in the remaining list.

- Time Complexity

    - Best Case: *O(N<sup>2</sup>)*
    - Worst Case: *O(N<sup>2</sup>)*
    - Average Case: *O(N<sup>2</sup>)*


- Space Complexity: O(1)

```python
def selection_sort(lst):
    size= len(lst)
    for i in range(size):
        min=i
        for j in range(i+1, size):
            if lst[j]<lst[min]:
                min= j
                swap= True
        lst[i], lst[min]= lst[min], lst[i]
    return lst
```

```python
lst= [8,6,4,2,7,9]
sorted_lst= selection_sort(lst)
print(sorted_lst)
```

## Insertion Sort

- Compare the current element with minimum element  in the remaining list.

- Time Complexity

    - Best Case: *O(N)*
    - Worst Case: *O(N<sup>2</sup>)*
    - Average Case: *O(N<sup>2</sup>)*


- Space Complexity: O(1)

```python
def insertion_sort_swap(lst):
    n= len(lst)
    for i in range(n-1):
        for j in range(i+1, 0, -1):
            if lst[j-1] > lst[j]:
                lst[j], lst[j-1] = lst[j-1], lst[j] # swap
            else:
                break
    return lst
```

```python
lst= [8,6,4,2,7,9]
sorted_lst= insertion_sort_swap(lst)
print(sorted_lst)
```

```python
def insertion_sort_shift(lst):
    n= len(lst)
    for i in range(1,n):
        key= lst[i]
        j= i-1
        while j>=0 and lst[j]>key:
            lst[j+1]= lst[j]
            j-=1
        lst[j+1]=key
    return lst
```

```python
def insertion_sort_shift_non_increasing(lst):
    n= len(lst)
    for i in range(1,n):
        key= lst[i]
        j= i-1
        while j>=0 and lst[j]<key:
            lst[j+1]= lst[j]
            j-=1
        lst[j+1]=key
    return lst
```

## Merge Sort

<center></center>

- Time Complexity

    - Best Case: *O(Nlog(N))*
    - Worst Case: *O(Nlog(N))*
    - Average Case: *O(Nlog(N))*


- Space Complexity: O(N)

```python
# Step1: Divide - O(1)
# Step2: Conquer - k*T(n/2)
#    here, k - depth of our recursive tree
#          T(n/2) - time taken for the subarray
# Step3: COMBINE - O(n)

def merge(arr, p, q, r): # Step3: COMBINE - O(n)
    l1= arr[p:q]
    l2= arr[q:r]
    l1.append(float('inf'))
    l2.append(float('inf'))
    i,j=0,0
    for rng in range(p,r):
        if l1[i]<=l2[j]:
            arr[rng]= l1[i]
            i+=1
        else:
            arr[rng]= l2[j]
            j+=1

def merge_sort(arr, p, r): # Step2: CONQUER - K*T(n/2)
    if r-p>1:
        q= int((r+p)/2) # Step 1: DIVIDE - O(1)
        merge_sort(arr, p, q)
        merge_sort(arr, q, r)
        merge(arr, p, q, r)

arr= [2,9,8,7,6,4,5,3]
size= len(arr)
merge_sort(arr, 0, size)
print(arr)
```
