# Linear Search

```python
import numpy as np

def linear_search(arr, key):
    for idx in range(len(arr)):
        if arr[idx]==key:
            return f"The key is  found at index {idx}"
    return f"The key is not found anywhere"

arr = np.random.randint(1,20,10)
key= 7
linear_search(arr, key)
```

# Binary Search

```python
def merge(arr, p, q, r):
    arr1 = arr[p:q]
    arr2 = arr[q:r]
    arr1.append(float("inf"))
    arr2.append(float("inf"))
    i,j= 0,0
    for idx in range(p,r):
        if arr1[i]<=arr2[j]:
            arr[idx] = arr1[i]
            i+=1
        else:
            arr[idx] = arr2[j]
            j+=1
    return arr

def merge_sort(arr, p, r):
    if r-p>1:
        q= int((r+p)/2)
        merge_sort(arr, p, q)
        merge_sort(arr, q, r)
        merge(arr, p, q, r)
    return arr

def binary_search(arr,key, p, r):
    if p>=r:
        print(f"Key nowhere to be found")
        return
    q= int((r+p)/2)

    if arr[q] == key:
        print(f"Key found at index {q+1}")
    elif arr[q]<key:
        binary_search(arr, key, q,r)
    else:
        binary_search(arr, key, p, q)

def binary(arr, key):
    print("Unsorted Array :", arr)
    merge_sort(arr, 0, len(arr))
    print("Sorted Array   :", arr)
    binary_search(arr, key, 0, len(arr))

arr= [2,9,8,7,6,4,5,3]
key = 5
binary(arr, key)
```
